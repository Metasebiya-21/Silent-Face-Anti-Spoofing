# 2 · Validate: pick the best checkpoint

Scores **every** checkpoint of one training run on the **validation** split and ranks them by ACER.
Use the validation split for choosing; keep the test split untouched until **03_test**.

```
datasets/rgb_image/val/<patch_info>/{0,1,2}/*.png
```
- **APCER** — attacks accepted as real (lower = safer)
- **BPCER** — real faces rejected (lower = friendlier)
- **ACER** — their average; the ranking key

## Parameters

````python
from model_development.config import REPO_ROOT

PATCH_INFO = "2.7_80x80"
SNAPSHOT_DIR = REPO_ROOT / "saved_logs" / "snapshot" / f"Anti_Spoofing_{PATCH_INFO}"
VAL_DIR = REPO_ROOT / "datasets" / "rgb_image" / "val" / PATCH_INFO
DEVICE = "cpu"   # or "cuda:0"
````

## Score every checkpoint

````python
from model_development.evaluation import evaluate_checkpoints

def show(results):
    """Print metrics as a table: lower APCER/BPCER/ACER is better."""
    print(f"{'checkpoint':58s} {'n':>5s} {'acc':>6s} {'APCER':>6s} {'BPCER':>6s} {'ACER':>6s}")
    for r in results:
        print(f"{r['checkpoint']:58s} {r['samples']:5d} {r['accuracy']:6.3f} {r['apcer']:6.3f} {r['bpcer']:6.3f} {r['acer']:6.3f}")

results = evaluate_checkpoints(SNAPSHOT_DIR, VAL_DIR, DEVICE)
assert results, f"no checkpoints in {SNAPSHOT_DIR}"
show(results)
````

## Best checkpoint

Copy this name into `CHECKPOINT` in **03_test**.

````python
best = results[0]
print("best:", best["checkpoint"])
print("confusion matrix (rows: true 0/1/2, columns: predicted 0/1/2):")
for row in best["confusion_matrix"]:
    print("  ", row)
````
